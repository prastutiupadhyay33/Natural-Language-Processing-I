## Question 1
**What is tokenization? Tokenize the sentence 'Python is easy to learn.' with spaCy, and 'Python is a popular programming language' with NLTK's word_tokenize.**

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
text="Python is easy to learn."
doc=nlp(text)
for token in doc:
  print(token.text)
```

```python
import nltk
from nltk.tokenize import word_tokenize
text="Python is a popular programming language"
tokens=word_tokenize(text)
print(tokens)
```

## Question 9
**Convert NLTK tokens to lowercase, and then write one complete program: sentences -> tokens -> remove punctuation -> lowercase -> count.**

```python
from nltk.tokenize import word_tokenize
text= "Python Python PYTHON is easy."
tokens=word_tokenize(text)
lower_tokens=[word.lower() for word in tokens]
print("Original tokens:")
print(tokens)
print("\nLowercase tokens:")
print(lower_tokens)
```

```python
import nltk
import string
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text="""
Python is a powerful programming language.
It is widely used in Data Science, Machine Learning, and NLP!
"""
sentences=sent_tokenize(text)
print("Sentences:")
for sentence in sentences:
  print(sentence)

tokens=word_tokenize(text)
print("\nAll Tokens:")
print(tokens)

words=[word for word in tokens if word not in string.punctuation]
print("\nTokens without punctuation:")
print(words)

lower_words=[word.lower() for word in words]
print("\nLowercase Tokens:")
print(lower_words)

print("\nTotal tokens:",len(tokens))
print("Total words:",len(words))
```
