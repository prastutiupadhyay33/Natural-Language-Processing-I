## Question 1
**What is stemming? Import NLTK's PorterStemmer, create the stemmer and stem the words playing, played, plays, player.**

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["playing","played","plays","player"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

## Question 2
**Stem every word of the sentence 'The students are studying different subjects.' using split().**

```python
text="The students are studying different subjects."
woerds=text.split()
for word in woerds:
    print(word,"->",stemmer.stem(word))
```

## Question 9
**Use the re module (regular expression) to pick the words from a sentence and stem them.**

```python
import re
text="Students are learning Python and developing machine learning projects."
words=re.findall(r'\b[a-zA-Z]+\b', text.lower())
for word in words:
    print(word,"->",stemmer.stem(word))
```

## Question 10
**Stem a full sentence and join the stemmed words back into one text. Print Original and Stemmed.**

```python
text="The players are playing football and practicing regularly."
words=text.lower().split()
stemmed_words=[]
for word in words:
    stemmed_words.append(stemmer.stem(word))
    stemmed_text=" ".join(stemmed_words)
print("Original:")
print(text)
print("\nStemmed:")
print(stemmed_text)
```
