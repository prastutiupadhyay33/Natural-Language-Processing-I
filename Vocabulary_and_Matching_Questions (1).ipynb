## Question 1
**What is spaCy's Matcher? Import spaCy and load the English model.**

```python
import spacy
nlp=spacy.load('en_core_web_sm')
```

## Question 2
**Import Matcher from spaCy and create a Matcher object.**

```python
from spacy.matcher import Matcher
matcher=Matcher(nlp.vocab)
```

## Question 3
**Create three patterns for 'solarpower', 'solar power' and 'solar-power'.**

```python
pattern1=[{'LOWER':'solarpower'}]
pattern2=[{'LOWER':'solar'},{'LOWER':'power'}]
pattern3=[{'LOWER':'solar'},{'IS_PUNCT':True},{'LOWER':'power'}]
```

## Question 4
**Add the three patterns to the matcher under the name 'SolarPower'.**

```python
matcher.add('SolarPower',[pattern1,pattern2,pattern3])
```

## Question 5
**Create a doc and find the matches in it.**

```python
doc=nlp(u"The solar Power Industry continues to grow as demand for solarpower increases. Solar-power cars are gaining popularity")
found_mataches=matcher(doc)
print(found_mataches)
```

## Question 6
**Print every match in a readable form (id, name, start, end and matched text).**

```python
for match_id,start,end in found_mataches:
    string_id=nlp.vocab.strings[match_id]
    span=doc[start:end]
    print(match_id,string_id,start,end,span.text)
```

## Question 7
**What is PhraseMatcher? Import it and create a PhraseMatcher object.**

```python
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)
```

## Question 8
**Read the text file 'reaganomics.txt' into a spaCy doc and print it.**

```python
intro="Reaganomics refers to the economic policies of President Ronald Reagan. Critics called it voodoo economics. Supporters described it as supply-side economics, which cut taxes to boost growth. It is also known as trickle-down economics. Reagan favored free-market economics over heavy regulation. "
filler="The government changed tax rules and many people discussed the effect on jobs and prices. "
with open('reaganomics.txt','w',encoding='cp1252') as f:
    f.write(intro+filler*200)
```

```python
with open('reaganomics.txt','r',encoding='cp1252')as f:
    doc3=nlp(f.read())
print(doc3[:100])
```

## Question 9
**Make a list of phrases and convert each phrase into a spaCy doc.**

```python
phrase_list=['voodoo economics','supply-side economics','trickle-down economics','free-market economics']
phrase_patterns=[nlp(text) for text in phrase_list]
```

## Question 10
**Add the phrase patterns to the PhraseMatcher and find the matches in the document.**

```python
matcher.add('VoodooEconomics',phrase_patterns)
matches=matcher(doc3)
matches
```
