## Question 1
**What is lemmatization? Load spaCy and find the lemma of every word in a simple sentence.**

```python
import spacy
nlp=spacy.load("en_core_web_sm")
text= "The boys are playing in the gardens."
doc=nlp(text)
for token in doc:
    print(token.text, "->", token.lemma_)
```

## Question 4
**Find the lemma of the irregular verb 'eat' (eating, ate, eaten). Skip punctuation.**

```python
text= "I am eating. She ate dinner. They have eaten food."
doc=nlp(text)
for token in doc:
  if token.is_alpha:
        print(token.text, "->", token.lemma_)
```

## Question 5
**Lemmatize a full sentence and join the lemmas into one text.**

```python
text= "The students were studying their lessons."
doc=nlp(text)
lemmatized_text= " ".join(token.lemma_ for token in doc)
print(lemmatized_text)
```

## Question 7
**Remove the punctuation, and then remove both stopwords and punctuation, and keep only the lemmas.**

```python
text= "Dogs are barking,cats are sleeping!"
doc=nlp(text)
words=[]
for token in doc:
  if not token.is_punct:
    words.append(token.lemma_)
print(words)

text= "The students are learning data science."
doc=nlp(text)
words=[]
for token in doc:
    if not token.is_stop and not token.is_punct:
        words.append(token.lemma_)
print(words)
```

## Question 8
**Create a function lemmatize_text() which removes stopwords and punctuation and returns the lemmatized text.**

```python
def lemmatize_text(text):
    doc=nlp(text)
    result=[]
    for token in doc:
        if not token.is_stop and not token.is_punct:
            result.append(token.lemma_)
    return " ".join(result)
text="The students were studying different subjects."
result=lemmatize_text(text)
print(result)
```

## Question 9
**Lemmatization with NLTK: install NLTK, download WordNet and lemmatize nouns and verbs.**

```python
!pip install nltk
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")

from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()

words=["cats","dogs","cars","houses"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word))

words=["playing","played","plays"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word,pos="v"))
```

## Question 10
**With NLTK, lemmatize irregular plurals, adjectives (pos='a') and adverbs (pos='r').**

```python
words=["children","mice","geese","men","women"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word,pos="n"))

words=["better","best","faster","happier"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word,pos="a"))

words=["quickly","slowly","happily"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word,pos="r"))
```
