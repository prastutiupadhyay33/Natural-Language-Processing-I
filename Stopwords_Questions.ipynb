## Question 1
**What are stopwords? Use spaCy to check stopwords, remove them from text, view spaCy's stopword list, add a custom stopword and create a `remove_stopwords()` function.**

```python
# Stopwords = very common words (the, is, and ...) that carry little meaning.
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
# Check whether single words are stopwords (is_stop gives True/False)
words=["the","is","Python","and","beautiful"]

for word in words:
  doc=nlp(word)
  for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
# Check every word of a sentence
words="Python is a very useful programming language."

doc=nlp(words)

for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
# Remove stopwords (keep only the words which are NOT stopwords)
text="This is a simple example of stopword removal."
doc=nlp(text)
stopwords=[]

for token in doc:
  if not token.is_stop:
     stopwords.append(token.text)
print(stopwords)
```

```python
# Remove stopwords AND punctuation
text="Python is an easy and powerful language."

doc=nlp(text)
words=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
   words.append(token.text)

print(words)
```

```python
# Print the result as a sentence - with punctuation
text="The students are learning data science in the classroom."

doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop:
   result.append(token.text)

print(" ".join(result))
```

```python
# Same sentence - without punctuation
text="The students are learning data science in the classroom."

doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
   result.append(token.text)

print(" ".join(result))
```

```python
# Convert the words to lowercase after removing stopwords
text="Python Is a Powerful Programming Language."
doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
     result.append(token.text.lower())

print(" ".join(result))
```

```python
# Display spaCy's stopword list
print(nlp.Defaults.stop_words)
```

```python
# Count how many stopwords spaCy has
print(len(nlp.Defaults.stop_words))
```

```python
# Add our own custom stopword "python"
nlp.Defaults.stop_words.add("python")
doc=nlp("Python is a popular programming language.")
for token in doc:
    print(token.text,"->",token.is_stop)
```

```python
# Function to remove stopwords and punctuation
def remove_stopwords(text):
  doc=nlp(text)
  result=[]
  for token in doc:
    if not token.is_stop and not token.is_punct:
     result.append(token.text)
  return " ".join(result)

text="The students are studying Natural Language Processing."
print(remove_stopwords(text))
```

## Question 2
**Do the same stopword tasks with NLTK: load the stopword list, check words, find and remove stopwords from tokenized text, count them and list the supported languages.**

```python
# Install and import nltk, then download the stopwords data
import nltk
nltk.download("stopwords")

from nltk.corpus import stopwords
# set() makes searching faster
stop_words=set(stopwords.words("english"))
```

```python
# How many English stopwords are there in NLTK?
stop_words=(stopwords.words("english"))
print(len(stop_words))
```

```python
# Check words in the stopword list (lower() so "Python" and "python" are treated the same)
words=["the","is","Python","and","beautiful"]
for word in words:
    print(word,"->",word.lower() in stop_words)
```

```python
# Tokenize a sentence and FIND the stopwords in it
from nltk.tokenize import word_tokenize
nltk.download("punkt")
nltk.download("punkt_tab")
text="Python is a very useful programming language."
tokens=word_tokenize(text)
result=[]
for word in tokens:
  if word.lower() in stop_words:
    result.append(word)
print(result)
```

```python
# Remove stopwords and punctuation
import string
stop_words=set(stopwords.words("english"))
text="The students are learning data science!"
tokens=word_tokenize(text)
result=[]
for word in tokens:
  # string.punctuation contains all punctuation marks
  if word.lower() not in stop_words and word not in string.punctuation:
      result.append(word)
print(result)
```

```python
# Remove stopwords and convert the words to lowercase
text="Python Is a Powerful Programming Language."
tokens=word_tokenize(text)
result=[]
for word in tokens:
  if word.lower() not in stop_words:
      result.append(word.lower())
print(result)
```

```python
# Count the number of stopwords in a sentence
text="This is a simple example to demonstrate stopword removal."
tokens=word_tokenize(text)
count=0
for word in tokens:
  if word.lower() in stop_words:
    count+=1
print("Number of stopwords:",count)
```

```python
# Languages which have a stopword list in NLTK
languages=stopwords.fileids()
print(languages)
```
