## Question 1
**What are stopwords? Load spaCy and check whether single words, and the words of a sentence, are stopwords using is_stop.**

```python
import spacy
nlp=spacy.load("en_core_web_sm")

words=["the","is","Python","and","beautiful"]

for word in words:
  doc=nlp(word)
  for token in doc:
    print(token.text,"->",token.is_stop)
words="Python is a very useful programming language."

doc=nlp(words)

for token in doc:
    print(token.text,"->",token.is_stop)
```

## Question 2
**Remove the stopwords from a sentence using spaCy.**

```python
text="This is a simple example of stopword removal."
doc=nlp(text)
stopwords=[]

for token in doc:
  if not token.is_stop:
     stopwords.append(token.text)
print(stopwords)
```

## Question 3
**Remove the stopwords and the punctuation from a sentence.**

```python
text="Python is an easy and powerful language."

doc=nlp(text)
words=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
   words.append(token.text)

print(words)
```

## Question 4
**Remove the stopwords and print the result as a sentence: first with punctuation, then without punctuation.**

```python
text="The students are learning data science in the classroom."

doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop:
   result.append(token.text)

print(" ".join(result))

text="The students are learning data science in the classroom."

doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
   result.append(token.text)

print(" ".join(result))
```

## Question 5
**Remove the stopwords and punctuation and convert the remaining words to lowercase.**

```python
text="Python Is a Powerful Programming Language."
doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
     result.append(token.text.lower())

print(" ".join(result))
```

## Question 8
**Create a function remove_stopwords() and test it on a sentence.**

```python
def remove_stopwords(text):
  doc=nlp(text)
  result=[]
  for token in doc:
    if not token.is_stop and not token.is_punct:
     result.append(token.text)
  return " ".join(result)

text="The students are studying Natural Language Processing."
print(remove_stopwords(text))
```

## Question 10
**With NLTK, count the stopwords in a sentence and display the languages which have a stopword list.**

```python
import nltk
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words=set(stopwords.words("english"))

text="This is a simple example to demonstrate stopword removal."

tokens=word_tokenize(text)
count=0
for word in tokens:
  if word.lower() in stop_words:
    count+=1
print("Number of stopwords:",count)

import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
languages=stopwords.fileids()
print(languages)
```
