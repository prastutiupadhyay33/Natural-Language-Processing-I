```python
import nltk
import spacy
import string
from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

nlp=spacy.load('en_core_web_sm')
```

```python
# Question 1: Read the text file and count its lines, words and characters.
# Explanation: open() with 'r' mode reads the file; read() gives the whole text and readlines() gives a list of lines.
with open('maple_street_story.txt','r',encoding='utf-8') as file:
    text=file.read()

with open('maple_street_story.txt','r',encoding='utf-8') as file:
    lines=file.readlines()

print(text[:300])
print()
for line in lines[:3]:
    print(line.strip())

print('Number of lines:',len(lines))
print('Number of words:',len(text.split()))
print('Number of characters:',len(text))
```

```python
# Question 2: Split the story into sentences using NLTK.
# Explanation: Sentence tokenization breaks a text into separate sentences.
sentences=sent_tokenize(text)
print('Number of sentences:',len(sentences))

for i,sentence in enumerate(sentences[:5],start=1):
    print(f'Sentence {i}:',sentence)
```

```python
# Question 3: Split the story into words, remove punctuation and convert to lowercase.
# Explanation: Word tokenization breaks text into tokens; isalpha() keeps only real words.
tokens=word_tokenize(text)
print('First 30 tokens:',tokens[:30])

words=[token for token in tokens if token.isalpha()]
lower_words=[word.lower() for word in words]

print('Total tokens:',len(tokens))
print('Total words:',len(words))
print('First 20 lowercase words:',lower_words[:20])
```

```python
# Question 4: Tokenize the story using spaCy.
# Explanation: spaCy gives every token properties such as is_alpha and is_punct, and doc.sents gives sentences.
doc=nlp(text)

for token in doc[:20]:
    print(token.text,token.is_alpha,token.is_punct)

print('Total tokens:',len(doc))
print('Total sentences:',len(list(doc.sents)))
```

```python
# Question 5: Remove stopwords using the NLTK stopword list.
# Explanation: Stopwords are common words like 'the' and 'is' that carry little meaning.
stop_words=set(stopwords.words('english'))

count=0
for word in words:
    if word.lower() in stop_words:
        count+=1

clean_words=[word.lower() for word in words if word.lower() not in stop_words]

print('Number of stopwords:',count)
print('Words after stopword removal:',len(clean_words))
print(clean_words[:20])
```

```python
# Question 6: Remove stopwords and punctuation using spaCy.
# Explanation: spaCy marks stopwords with is_stop and punctuation with is_punct.
spacy_words=[]

for token in doc:
    if not token.is_stop and not token.is_punct and not token.is_space:
        spacy_words.append(token.text.lower())

print(spacy_words[:20])
print(Counter(spacy_words).most_common(5))
```

```python
# Question 7: Apply stemming using the NLTK PorterStemmer.
# Explanation: Stemming cuts a word down to its root form by removing endings (it may not be a real word).
stemmer=PorterStemmer()

for word in clean_words[:20]:
    print(word,'->',stemmer.stem(word))
```

```python
# Question 8: Apply lemmatization using spaCy.
# Explanation: Lemmatization changes a word to its dictionary base form, e.g. 'playing' becomes 'play'.
count=0
for token in doc:
    if token.is_alpha:
        print(token.text,'->',token.lemma_)
        count+=1
    if count==20:
        break
```

```python
# Question 9: Apply lemmatization using the NLTK WordNetLemmatizer.
# Explanation: The result depends on the part of speech: 'n' for noun and 'v' for verb.
lemmatizer=WordNetLemmatizer()

for word in clean_words[:15]:
    print(word,'-> noun:',lemmatizer.lemmatize(word,pos='n'),'| verb:',lemmatizer.lemmatize(word,pos='v'))
```

```python
# Question 10: Do POS tagging using spaCy and count the tags.
# Explanation: POS tagging gives each word a grammar label such as noun, verb or adjective.
for token in doc[:20]:
    print(token.text,token.pos_,spacy.explain(token.pos_))

pos_counts=Counter(token.pos_ for token in doc)
print(pos_counts)
```

```python
# Question 11: Do POS tagging using NLTK and list the nouns, verbs and adjectives.
# Explanation: NLTK tags starting with NN are nouns, VB are verbs and JJ are adjectives.
tags=nltk.pos_tag(words[:100])

print('Nouns:',[word for word,tag in tags if tag.startswith('NN')])
print('Verbs:',[word for word,tag in tags if tag.startswith('VB')])
print('Adjectives:',[word for word,tag in tags if tag.startswith('JJ')])
```

```python
# Question 12: Clean the story (stopwords, punctuation, lemma), save it to a new file and search a word in it.
# Explanation: This combines the earlier steps and shows how to write to a file ('w') and read it back.
result=[]
for token in doc:
    if not token.is_stop and not token.is_punct and not token.is_space:
        result.append(token.lemma_.lower())

with open('cleaned_story.txt','w',encoding='utf-8') as file:
    file.write(' '.join(result))

with open('cleaned_story.txt','r',encoding='utf-8') as file:
    cleaned=file.read()

print(cleaned[:200])
print('Number of words in cleaned file:',len(cleaned.split()))
print('Occurrences of lamp:',cleaned.count('lamp'))
```
